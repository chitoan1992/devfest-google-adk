````python
# @title 🎨 INITIALIZE GAME SYSTEM
import datetime
import os
from zoneinfo import ZoneInfo
from google.adk.agents import Agent
from google.adk.sessions import InMemorySessionService
from google.adk.runners import Runner
from google.genai import types
from IPython.display import HTML, display
import warnings
import logging

# Suppress warnings
warnings.filterwarnings("ignore")
logging.basicConfig(level=logging.CRITICAL)
logging.getLogger('asyncio').setLevel(logging.CRITICAL)
logging.getLogger('google_genai').setLevel(logging.CRITICAL)
logging.getLogger('google.adk').setLevel(logging.CRITICAL)

# ═══════════════════════════════════════════════════════════
# GAMIFICATION SYSTEM
# ═══════════════════════════════════════════════════════════

class GameSystem:
    """🎮 Gamification and Progress Tracking System"""
    
    def __init__(self):
        self.xp = 0
        self.level = 0
        self.achievements = []
        self.missions_completed = []
        
        self.levels = {
            0: {"name": "🔰 INITIATE", "xp_required": 0},
            1: {"name": "⚙️  APPRENTICE", "xp_required": 1000},
            2: {"name": "👥 COORDINATOR", "xp_required": 2500},
            3: {"name": "🚀 ARCHITECT", "xp_required": 5000},
            4: {"name": "🧠 STRATEGIST", "xp_required": 7500},
            5: {"name": "🎯 MASTER BUILDER", "xp_required": 10000}
        }
        
    def add_xp(self, amount, reason=""):
        """Add XP and check for level up"""
        self.xp += amount
        old_level = self.level
        
        # Check for level up
        for level, data in sorted(self.levels.items(), reverse=True):
            if self.xp >= data["xp_required"]:
                self.level = level
                break
        
        # Display XP gain
        print(f"\n⚡ +{amount} XP - {reason}")
        
        # Level up animation
        if self.level > old_level:
            self._level_up_animation()
    
    def _level_up_animation(self):
        """Display level up animation"""
        level_data = self.levels[self.level]
        print("""
╔═══════════════════════════════════════════════════════╗
║                                                       ║
║           ⭐⭐⭐ LEVEL UP! ⭐⭐⭐                        ║
║                                                       ║
""")
        print(f"║     You are now: {level_data['name']:^30}      ║")
        print("""
║                                                       ║
╚═══════════════════════════════════════════════════════╝
""")
    
    def add_achievement(self, achievement_name, xp_reward=0):
        """Unlock achievement"""
        if achievement_name not in self.achievements:
            self.achievements.append(achievement_name)
            print(f"\n🏆 ACHIEVEMENT UNLOCKED: {achievement_name}")
            if xp_reward > 0:
                self.add_xp(xp_reward, f"Achievement: {achievement_name}")
    
    def complete_mission(self, mission_name, xp_reward):
        """Mark mission as completed"""
        if mission_name not in self.missions_completed:
            self.missions_completed.append(mission_name)
            print(f"\n✅ MISSION COMPLETE: {mission_name}")
            self.add_xp(xp_reward, f"Mission: {mission_name}")
    
    def show_stats(self):
        """Display current player stats"""
        level_data = self.levels[self.level]
        next_level_data = self.levels.get(self.level + 1, {"xp_required": self.xp})
        xp_to_next = next_level_data["xp_required"] - self.xp
        
        # Progress bar
        if self.level < 5:
            progress = (self.xp - level_data["xp_required"]) / (next_level_data["xp_required"] - level_data["xp_required"])
            bar_length = 30
            filled = int(bar_length * progress)
            bar = "█" * filled + "░" * (bar_length - filled)
        else:
            bar = "█" * 30
        
        print(f"""
┌─────────────────────────────────────────────────────────┐
│  📊 YOUR STATS                                          │
│  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━  │
│  Level: {self.level} - {level_data['name']:20}              │
│  XP: {self.xp:,} / {next_level_data['xp_required']:,}                                   │
│  [{bar}] {int(progress * 100) if self.level < 5 else 100}%        │
│  Achievements: {len(self.achievements)} / 11                               │
│  Missions: {len(self.missions_completed)} / 6                                  │
└─────────────────────────────────────────────────────────┘
""")

# Initialize game system
game = GameSystem()

# ═══════════════════════════════════════════════════════════
# COLOR SYSTEM (Cyberpunk Theme)
# ═══════════════════════════════════════════════════════════

class Colors:
    """Cyberpunk color palette"""
    NEON_PINK = '\033[95m'
    CYBER_BLUE = '\033[94m'
    NEON_CYAN = '\033[96m'
    MATRIX_GREEN = '\033[92m'
    WARNING_YELLOW = '\033[93m'
    ERROR_RED = '\033[91m'
    RESET = '\033[0m'
    BOLD = '\033[1m'
    DIM = '\033[2m'

def print_success(text):
    print(f"{Colors.MATRIX_GREEN}{text}{Colors.RESET}")

def print_info(text):
    print(f"{Colors.NEON_CYAN}{text}{Colors.RESET}")

def print_warning(text):
    print(f"{Colors.WARNING_YELLOW}{text}{Colors.RESET}")

def print_error(text):
    print(f"{Colors.ERROR_RED}{text}{Colors.RESET}")

def print_header(text):
    print(f"{Colors.BOLD}{Colors.NEON_PINK}{text}{Colors.RESET}")

# ═══════════════════════════════════════════════════════════
# INITIALIZATION COMPLETE
# ═══════════════════════════════════════════════════════════

print("""
╔═══════════════════════════════════════════════════════════╗
║                                                           ║
║           ✅ SYSTEM INITIALIZATION COMPLETE               ║
║                                                           ║
║  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━ ║
║  ✓ Game System: LOADED                                    ║
║  ✓ Color System: ACTIVE                                   ║
║  ✓ ADK Framework: READY                                   ║
║  ✓ Progress Tracking: ONLINE                              ║
║                                                           ║
╚═══════════════════════════════════════════════════════════╝
""")

game.add_xp(200, "Mission 0: System Initialization")
game.add_achievement("🔧 System Administrator", 100)
game.show_stats()

print_success("\n🚀 Ready to build agents with Gemini 2.5 Flash!")
print_info("\n💡 Tip: Complete all missions to reach level 5 and unlock Grand Master achievement!")
````

**💡 Critical Intel:** 
```
DOCSTRING = AI's Instruction Manual
```

The LLM reads your docstring to understand:
- ❓ **WHAT** does this tool do?
- ⚡ **WHEN** should I use it?
- 📋 **WHAT** parameters does it need?
- 🎁 **WHAT** does it return?

````python
# @title 🔨 TOOL CREATION: get_weather()

print("""
╔════════════════════════════════════════════════════════╗
║  🔨 INITIATING TOOL FORGE PROTOCOL                     ║
║  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━ ║
║  Crafting weapon: get_weather()                        ║
║  Type: Data Retrieval Tool                             ║
║  Status: Manufacturing...                              ║
╚════════════════════════════════════════════════════════╝
""")

def get_weather(city: str) -> dict:
    """Lấy báo cáo thời tiết hiện tại cho một thành phố.
    
    Args:
        city: Tên thành phố (VD: "Hà Nội", "London", "Tokyo")
    
    Returns:
        dict: {'status': 'success'|'error', 'report': str} hoặc {'error_message': str}
    """
    print_info(f"🔍 Đang tra cứu thời tiết: {Colors.BOLD}{city}{Colors.RESET}")
    
    city_normalized = city.lower().replace(" ", "")
    
    # Mock weather database
    mock_weather_db = {
        "hanoi": {"status": "success", "report": "☀️ Hà Nội: Nắng đẹp, 25°C"},
        "hànội": {"status": "success", "report": "☀️ Hà Nội: Nắng đẹp, 25°C"},
        "danang": {"status": "success", "report": "☀️ Đà Nẵng: Nắng đẹp, 27°C"},
        "đànẵng": {"status": "success", "report": "☀️ Đà Nẵng: Nắng đẹp, 27°C"},
        "london": {"status": "success", "report": "☁️ London: Nhiều mây, 15°C"},
        "tokyo": {"status": "success", "report": "🌧️ Tokyo: Mưa nhẹ, 18°C"},
    }
    
    if city_normalized in mock_weather_db:
        result = mock_weather_db[city_normalized]
        print_success(f"✅ Tìm thấy dữ liệu thời tiết!")
        return result
    else:
        print_error(f"❌ Không có dữ liệu cho '{city}'")
        return {"status": "error", "error_message": f"Xin lỗi, không có thông tin thời tiết cho '{city}'."}

# Test tool
print("""
╔════════════════════════════════════════════════════════╗
║  ✅ TOOL FORGED SUCCESSFULLY!                          ║
║  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━ ║
║  Running diagnostic tests...                           ║
╚════════════════════════════════════════════════════════╝
""")

print_header("\n📝 TEST SEQUENCE:")
print(get_weather("Hà Nội"))
print(get_weather("Đà Nẵng"))
print(get_weather("Unknown City"))

print("""
╔════════════════════════════════════════════════════════╗
║  🎉 DIAGNOSTIC COMPLETE!                               ║
╚════════════════════════════════════════════════════════╝
""")

game.add_xp(300, "Phase 1: Tool Creation")
game.add_achievement("🔧 Tool Smith", 0)
````

![](https://raw.githubusercontent.com/chitoan1992/devfest-google-adk/refs/heads/master/media/Weather%20Agent%20Flow.svg)

```
╔═══════════════════════════════════════════════════════════╗
║  📡 ASYNC EVENT FLOW                                      ║
║  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━ ║
║                                                           ║
║  1. User Query ──────> Runner                             ║
║  2. Runner ──────────> Agent (Analyze)                    ║
║  3. Agent ───────────> Tool (Execute)                     ║
║  4. Tool ────────────> Result                             ║
║  5. Agent ───────────> LLM (Generate Response)            ║
║  6. Runner ──────────> Final Response                     ║
║                                                           ║
║  Why Async? LLM & API calls are I/O-bound!                ║
║  Async = Better performance & responsiveness              ║
║                                                           ║
╚═══════════════════════════════════════════════════════════╝
```

````python
# @title 💬 INTERACTION FUNCTIONS

async def call_agent_async(query: str, runner, user_id, session_id):
    """Gửi query đến agent và hiển thị response."""
    print(f"\n{Colors.BOLD}{Colors.NEON_CYAN}{'═'*60}{Colors.RESET}")
    print(f"{Colors.BOLD}👤 User:{Colors.RESET} {Colors.CYBER_BLUE}{query}{Colors.RESET}")
    print(f"{Colors.BOLD}{Colors.NEON_CYAN}{'═'*60}{Colors.RESET}")
    
    content = types.Content(role='user', parts=[types.Part(text=query)])
    final_response_text = "⚠️ Agent không tạo ra response."
    
    async for event in runner.run_async(user_id=user_id, session_id=session_id, new_message=content):
        if event.is_final_response():
            if event.content and event.content.parts:
                final_response_text = event.content.parts[0].text
            elif event.actions and event.actions.escalate:
                final_response_text = f"⚠️ Agent escalated: {event.error_message or 'Không có message.'}"
            break
    
    print(f"\n{Colors.BOLD}🤖 Agent:{Colors.RESET} {Colors.MATRIX_GREEN}{final_response_text}{Colors.RESET}")
    print(f"{Colors.BOLD}{Colors.NEON_CYAN}{'═'*60}{Colors.RESET}\n")

print_success("✅ Interaction functions loaded!")
````

````python
# @title 🔨 CREATE SPECIALIST TOOLS
from typing import Optional

print("""
╔════════════════════════════════════════════════════════╗
║  🔨 FORGING SPECIALIST TOOLS                           ║
║  ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━ ║
╚════════════════════════════════════════════════════════╝
""")

def say_hello(name: Optional[str] = None) -> str:
    """Cung cấp lời chào thân thiện.
    
    Args:
        name: Tên người cần chào (optional)
    
    Returns:
        str: Message chào hỏi
    """
    if name:
        greeting = f"👋 Xin chào, {name}!"
        print_info(f"🔔 say_hello called with: {Colors.BOLD}{name}{Colors.RESET}")
    else:
        greeting = "👋 Xin chào!"
        print_info(f"🔔 say_hello called (no name)")
    return greeting

def say_goodbye() -> str:
    """Cung cấp message tạm biệt."""
    print_info(f"🔔 say_goodbye called")
    return "👋 Tạm biệt! Chúc bạn một ngày tốt lành! ✨"

# Test
print_header("\n📝 TOOL TESTS:")
print(say_hello("Commander"))
print(say_hello())
print(say_goodbye())

print("""
\n╔════════════════════════════════════════════════════════╗
║  ✅ SPECIALIST TOOLS READY!                            ║
╚════════════════════════════════════════════════════════╝
""")

game.add_xp(300, "Phase 1: Specialist Tools Created")
````

```
MVP ≠ Minimum Viable Product
MVP = Minimum VALUABLE Product

3-5 core features is enough.
```

## 🎯 PITCH STRUCTURE

```
Total: 135 seconds (2m15s) - Perfect for 3-5 min slot with Q&A buffer

🎬 Hook (15s)         → "Imagine..."
💔 Problem (30s)      → Real pain + stats
✨ Solution (45s)     → Product demo
```

**💡 Pro Tips:**
- Story > Features
- Passion > Polish (authenticity wins)
- Practice demo 10x before pitch
- Add demo video/GIF → conversion rate ↑↑↑
